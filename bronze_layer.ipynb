### Bronze Layer- Raw Data Ingestion

Loading the data

```python

df_trips = spark.read.format('csv') \
.option('header','true') \
.option('inferSchema','true')\
.load('/Volumes/taxi_catalog/taxi_project_schema/taxi_volume/taxi_trip_data_raw.csv')

df_trips.show(5)
```

```python
df_zones = spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load("/Volumes/taxi_catalog/taxi_project_schema/taxi_volume/taxi_zone_lookup.csv")

df_zones.show(5)
```

Storing the loaded data as Delta tables

```python
df_trips.write.format('delta').saveAsTable('taxi_catalog.taxi_project_schema.bronze_trips_table')
```

```python
df_zones.write.format('delta').saveAsTable('taxi_catalog.taxi_project_schema.bronze_zones_table')
```
