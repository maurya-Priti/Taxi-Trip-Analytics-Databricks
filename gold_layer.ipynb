### Gold Layer- Business Metrics

```python
%sql
CREATE OR REPLACE TABLE taxi_catalog.taxi_project_schema.gold_zone_revenue AS
SELECT
pickup_zone,
SUM(total_amount) AS total_revenue,
COUNT(*) AS total_trips
FROM taxi_catalog.taxi_project_schema.silver_trips_enriched
GROUP BY pickup_zone
ORDER BY total_revenue DESC
```

```python
%sql
SELECT 
pickup_zone,
pickup_latitude,
pickup_longitude,
Sum(total_amount) AS total_revenue
FROM taxi_catalog.taxi_project_schema.silver_trips_enriched
GROUP BY 
pickup_zone,
pickup_latitude,
pickup_longitude
LIMIT 10;
```

Revenue by time of the day

```python
%sql
CREATE OR REPLACE TABLE taxi_catalog.taxi_project_schema.gold_time_of_day_revenue AS
SELECT 
CASE
    WHEN pickup_hour BETWEEN 0 AND 6 THEN 'Midnight'
    WHEN pickup_hour BETWEEN 7 AND 11 THEN 'Morning'
    WHEN pickup_hour BETWEEN 12 AND 16 THEN 'Noon'
    WHEN pickup_hour BETWEEN 17 AND 21 THEN 'Evening'
    ELSE 'Night'
END AS time_of_day,
SUM(total_amount) AS total_revenue,
COUNT(*) AS total_trips
FROM taxi_catalog.taxi_project_schema.silver_trips_enriched
GROUP BY time_of_day
```
